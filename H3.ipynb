```python
def get_age_group(age: float) -> str:
  if age < 12: return "Child"
  elif 12 <= age <= 18: return "Teen"
  else: return "Adult"

dataframe["AgeGroup"] = dataframe["Age"].apply(get_age_group)

dataframe
```

```python
def get_fare_category(fare: float) -> "str":
  if fare < 20.0: return "Low"
  elif 20.0 <= fare <= 50.0: return "Medium"
  else: return "High"

dataframe["FareCategory"] = dataframe["Fare"].apply(get_fare_category)

dataframe["FareCategory"].value_counts()
```
